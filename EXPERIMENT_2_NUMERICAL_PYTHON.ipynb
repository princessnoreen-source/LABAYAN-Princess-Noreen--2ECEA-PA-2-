```python
import numpy as np
```

### **A. REPRODUCIBLE NORMALIZATION PROBLEM**

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
Z = (X - np.mean(X)) / (np.std(X))
X_normalized = Z

np.save('X normalized.npy', X_normalized)
```

```python
print ("X array: \n", X)
print ("\n X_normalized: \n", X_normalized)
print ("\n X_normalized mean: ", (round (np.mean (X_normalized))))
print ("\n X_normalized standard deviation: ", (round(np.std (X_normalized))))
```

### **B. CUBES DIVISIBLE BY 4 PROBLEM**

```python
c = np.arange (1, 101, 1)
C = np.power (c,3)
C = C.reshape (10,10)
div_by_4 = C[C % 4 == 0]

np.save('div_by_4.npy', div_by_4)
```

```python
print ("Shape of C array: ",  C.shape)
print ("\n div_by_4 array: \n",  div_by_4)
print ("\n Number of Values in div_by_4: ", len(div_by_4))
```

### **C. ABOVE-MEAN SQUARES PROBLEM**

```python
S = np.arange (1, 37, 1)
S = np.power (S,2)
S = S.reshape (6,6)
S_mean = np.mean(S)
above_mean = S[S > S_mean ]

np.save('above_mean', above_mean)
```

```python
print ("S array \n", S, "\n")
print ("Mean of S array: \n ", S_mean, "\n")
print ("Above mean array: \n", above_mean, "\n")
print ("Number of above mean values: ", len(above_mean))
```
